# RiskPulse DeepML — 02. Data Validation & Outlier Handling

**Objective**:
- Remove duplicate records from the dataset.
- Validate demographic features (`person_age`, `person_emp_length`).
- Enforce positive financial bounds (`person_income`, `loan_amnt`).
- Interest Rate & Ratio Consistency Checks
- Export the cleaned dataset to `data/cleaned_credit_risk_dataset.csv` for feature engineering and model training.

### Environment Setup & Data Loading


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual styling for outlier inspection
sns.set_theme(style="whitegrid", palette="muted")
%matplotlib inline

# 1. Load the raw credit risk dataset
raw_data_path = '../data/credit_risk_dataset.csv'
df = pd.read_csv(raw_data_path)

# 2. Inspect initial dataset shape and structure
print(f"Raw Dataset Shape: {df.shape}")
print(f"Total Rows: {len(df):,}")
print(f"Total Columns: {df.shape[1]}")

df.head()

df_copy = df.copy()
```

###  Duplicate Detection & Removal

```python
df_copy.shape[0]
```

### Demographic Sanity Checks (`person_age` & `person_emp_length`) and Finanacial Bounds check (`loan_amnt`)

```python
#Remove ages below 18 or below 100
df_copy= df_copy[df_copy["person_age"] >= 18]
df_copy= df_copy[df_copy["person_age"] <= 100]

#Remove employment length greater than age or extreme outliers (>60)
df_copy = df_copy[df_copy["person_emp_length"] <= df_copy["person_age"]]
df_copy= df_copy[df_copy["person_emp_length"] <= 60]

# Financial Bounds Check
df_copy= df_copy[df_copy["person_income"] > 0]
```

### Interest Rate & Ratio Consistency Checks

```python
df_copy.describe()
```

```python
print(f"Loand interest rate minimum: {df_copy["loan_int_rate"].min()} and maximum: {df_copy["loan_int_rate"].max()}")
```

### Summary of Data Cleaning & Exporting Clean Dataset

```python
df_clean = df_copy
```

```python
# Path relative to the notebooks folder
ouput_path = '../data/cleaned_credit_risk_dataset.csv'

# Export cleaned dataframe without index numbers
df_clean.to_csv(ouput_path, index=False)
```
